### Fictional case study
The objective of this assessment is to use the skills you have developed in your first few practicals to address a specific research question with a specific analytical task. The setting is the (fictional) town of Coburg, South Carolina, U.S., which has a population of 35,000 people. This coastal town borders the ocean, and is divided into 6 administrative districts. All districts have roughly the same populations.

Three of the districts border the ocean and have beachfronts: the Ocean (district 1), Dunes (district 2) and North Beach (district 3). The other three, the Western (district 4), Grove (district 5) and Central (district 6), are inland and do not contain coastland. The inland and coastal districts differ further in that the inland districts have on average younger populations, a larger number of children per household, and more schools per district (there are a number of retiree, i.e. pensioner, communities in the coastal districts). Throughout this exercise and in the data itself, the districts are referred to by their numerical codes 1–6.

Recently there has been concern about the levels of crime (theft, vandalism and drug arrests) in the town, and claims have been made by members of the town council that rising crime levels are causing a drop in real estate value. A study has been commissioned on whether to increase the budget of the police department. A crucial question for the town authorities is whether to do this citywide or on a more district-by-district basis in order to stop or slow this trend.

### The data
Data on the number of crimes per thousand persons (in each year, in each district) for the period 2001–2010 has been compiled by police precincts in the districts. Additionally, records of average house selling price (by year and district) have been obtained from real estate brokers for the same period. The two datasets have been combined into a single data file, **HousePricesCrimeRates.csv**, which the code cell below loads for you as **`coburg`**. The file describes fictional crime statistics and home selling prices (in inflation-corrected US dollars) for each district in the 2001–2010 period.

There are 13 columns: the first column (`year`) is the year, the following 6 columns (`district1_HousePrice` to `district6_HousePrice`) give the average house selling price for each district, and the next 6 columns (`district1_CrimeRate` to `district6_CrimeRate`) give the number of crimes per 1,000 people in each district. Each row corresponds to a single year, so each value is the annual average house price or crime rate in a district for that year. For instance, the third row of `district2_HousePrice` is the average house price in district 2 in 2003.
